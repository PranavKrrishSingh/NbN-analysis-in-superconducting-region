```python
import numpy as np
import matplotlib.pyplot as plt

hbar = 1
m = 1
mu = 10
Delta = 1.5  # meV

# Momentum
k = np.linspace(-2*np.sqrt(2*mu), 2*np.sqrt(2*mu), 500)

# Quasiparticle energy
E_k = np.sqrt((hbar**2 * k**2 / (2*m) - mu)**2 + Delta**2)

plt.figure(figsize=(7,5))
plt.plot(k, E_k)
plt.xlabel("Momentum k")
plt.ylabel("Quasiparticle Energy E(k) [meV]")
plt.title("NbN Thin Film: Energy vs Momentum (s-wave BCS)")
plt.grid(True)
plt.show()
```
