```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
t = 1
t_prime = 0.3
mu = 1
Delta0 = 0.2 #mev

k_line = np.linspace(-np.pi, np.pi, 300)

#
xi_k = -2*t*(np.cos(k_line) + np.cos(k_line)) - 4*t_prime*np.cos(k_line)**2 - mu


Delta_chiral = Delta0 * np.sqrt(2) * np.sin(k_line)
E_chiral = np.sqrt(xi_k**2 + Delta_chiral**2)

Delta_helical = Delta0 * np.sqrt(2) * np.sin(k_line)
E_helical = np.sqrt(xi_k**2 + Delta_helical**2)


plt.figure(figsize=(8,5))
plt.plot(k_line, E_chiral, label="Chiral p-wave")
plt.plot(k_line, E_helical+0.1, '--', label="Helical p-wave")
#offset is added delibrately
plt.ylabel("Quasiparticle Energy E(k)")
plt.xlabel("Momentum k")
plt.title("Sr2RuO4 BCS Hamiltonian: Chiral vs Helical p-wave")
plt.legend()
plt.grid(True)
plt.show()


```

```python
import numpy as np
import matplotlib.pyplot as plt

hbar = 1
m = 1
mu = 10
Delta = 1.5  # meV

# Momentum
k = np.linspace(-2*np.sqrt(2*mu), 2*np.sqrt(2*mu), 500)

# Quasiparticle energy
E_k = np.sqrt((hbar**2 * k**2 / (2*m) - mu)**2 + Delta**2)

plt.figure(figsize=(7,5))
plt.plot(k, E_k)
plt.xlabel("Momentum k")
plt.ylabel("Quasiparticle Energy E(k) [meV]")
plt.title("NbN Thin Film: Energy vs Momentum (s-wave BCS)")
plt.grid(True)
plt.show()
```
